```python
# cols that should not be zero but are zero in the dataset
cols = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]
```

```python
# replacing the zero values with the median values of each column
for col in cols:
    diabetes_dataset[col] = diabetes_dataset[col].replace(0, diabetes_dataset[col].median())
```

```python
# separating the data and labels
X = diabetes_dataset.drop(columns = 'Outcome', axis=1)
Y = diabetes_dataset['Outcome']
```

Train Test Split

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.2, stratify=Y, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

Training the Model

```python
import numpy as np

from sklearn.base import clone
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# Use the original RFC configuration as the baseline
base_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=6,
    random_state=42,
    n_jobs=-1
)

# Generate out-of-fold probabilities using TRAINING DATA only
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

oof_prob = cross_val_predict(
    base_model,
    X_train,
    Y_train,
    cv=cv,
    method="predict_proba",
    n_jobs=-1
)[:, 1]

# Select threshold that maximizes out-of-fold F1
thresholds = np.arange(0.20, 0.71, 0.01)

scores = [
    f1_score(Y_train, (oof_prob >= t).astype(int))
    for t in thresholds
]

scores_array = np.asarray(scores, dtype=float)
best_idx = int(np.argmax(scores_array))

best_threshold = float(thresholds[best_idx])
best_f1 = scores[best_idx]

print("Selected threshold:", round(best_threshold, 2))
print("Training OOF F1:", (max(scores), 4))

# Fit on the complete training set
base_model.fit(X_train, Y_train)

# Evaluate once on held-out test data
test_prob = base_model.predict_proba(X_test)[:, 1]
test_pred = (test_prob >= best_threshold).astype(int)

print("\nTest Accuracy:", round(accuracy_score(Y_test, test_pred), 4))
print("Test Precision:", (precision_score(Y_test, test_pred, zero_division=0), 4))
print("Test Recall:", (recall_score(Y_test, test_pred, zero_division=0), 4))
print("Test F1:", (f1_score(Y_test, test_pred, zero_division=0), 4))

print("\nConfusion Matrix:\n", confusion_matrix(Y_test, test_pred))
print("\n", classification_report(Y_test, test_pred, zero_division=0))
```
